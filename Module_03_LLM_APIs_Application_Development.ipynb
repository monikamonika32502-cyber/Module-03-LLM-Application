```python

# DAY 11 - LLM API CONNECTION

from google import genai
from google.genai import types
from google.colab import userdata

# Load API key securely
api_key = userdata.get("GEMINI_API_KEY")

# Connect Gemini
client = genai.Client(api_key=api_key)

MODEL = "gemini-3.8-flash"

# Generate first response
response = client.models.generate_content(
    model=MODEL,
    contents="Explain Generative AI in simple words with 3 examples."
)

print("GEMINI API CONNECTED SUCCESSFULLY!")
print("\nAI RESPONSE:")
print(response.text)
```

```python

# DAY 12 - CONVERSATION AND CHAT HISTORY

from google.genai import types

# Use the model that worked in Day 11
MODEL = "gemini-3.8-flash"

# Store conversation history
history = []

def ask_ai(question):
    history.append(
        types.Content(
            role="user",
            parts=[types.Part(text=question)]
        )
    )

    response = client.models.generate_content(
        model=MODEL,
        contents=history
    )

    history.append(
        types.Content(
            role="model",
            parts=[types.Part(text=response.text)]
        )
    )

    print("AI:", response.text)

# Ask first question
print("QUESTION 1: What is Python?")
ask_ai("What is Python?")

# Ask second question
print("\nQUESTION 2: Applications of Python")
ask_ai("Give me three applications of Python.")

# Display conversation history
print("\n--- CHAT HISTORY ---")

for message in history:
    print(message.role.upper(), ":", message.parts[0].text)


```

# DAY 13 – STRUCTURED RESPONSES AND ERROR HANDLING

```python

# DAY 13 - STRUCTURED RESPONSES

import json
from google.genai import types

try:
    response = client.models.generate_content(
        model="gemini-flash-lite-latest",
        contents="Give information about Python programming language.",
        config=types.GenerateContentConfig(
            response_mime_type="application/json",
            response_schema={
                "type": "OBJECT",
                "properties": {
                    "language": {"type": "STRING"},
                    "creator": {"type": "STRING"},
                    "applications": {
                        "type": "ARRAY",
                        "items": {"type": "STRING"}
                    }
                },
                "required": ["language", "creator", "applications"]
            }
        )
    )

    result = json.loads(response.text)

    print("STRUCTURED AI RESPONSE:")
    print(json.dumps(result, indent=4))

except Exception as e:
    print("ERROR OCCURRED:", e)
```

# DAY 14 – AI CHAT APPLICATION

```python

# DAY 14 - SIMPLE AI CHATBOT

import ipywidgets as widgets
from IPython.display import display, clear_output

# Create chatbot interface
question_box = widgets.Text(
    placeholder="Ask me anything...",
    description="You:"
)

send_button = widgets.Button(
    description="Send",
    button_style="success"
)

output = widgets.Output()

# Chat history
chat_history = []

def send_message(button):
    question = question_box.value.strip()

    if not question:
        return

    with output:
        print("You:", question)

        try:
            response = client.models.generate_content(
                model="gemini-flash-lite-latest",
                contents=question
            )

            print("AI:", response.text)
            chat_history.append({
                "user": question,
                "assistant": response.text
            })

        except Exception as e:
            print("Error:", e)

    question_box.value = ""

send_button.on_click(send_message)

# Display chatbot
display(question_box, send_button, output)

print("AI CHATBOT READY!")
```

# DAY 15 – TESTING AND FINAL SUBMISSION

```python

# DAY 15 - FINAL TESTING

test_questions = [
    "What is Artificial Intelligence?",
    "What is Machine Learning?",
    "Explain Python in simple words."
]

print("FINAL CHATBOT TESTING")
print("=" * 40)

for i, question in enumerate(test_questions, 1):
    try:
        response = client.models.generate_content(
            model="gemini-flash-lite-latest",
            contents=question
        )

        print(f"\nTest {i}: {question}")
        print("AI Response:", response.text)
        print("Status: PASSED")

    except Exception as e:
        print(f"\nTest {i}: FAILED")
        print("Error:", e)

print("\nFINAL TESTING COMPLETED!")
```
